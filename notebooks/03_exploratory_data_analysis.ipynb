```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Clean Dataset Load Karein
df = pd.read_csv("../data/telco_customer_churn_clean.csv")
```

```python
# 1. Churn Counts & Percentages
churn_counts = df["Churn"].value_counts()
churn_pct = df["Churn"].value_counts(normalize=True) * 100

total_customers = len(df)
churned_customers = churn_counts.get("Yes", 0)
churn_rate = churn_pct.get("Yes", 0.0)

# 2. Print Summary Metrics
print(f"Total Customers: {total_customers}")
print(f"Churned Customers: {churned_customers}")
print(f"Churn Rate: {churn_rate:.2f}%")

# 3. Chart Generation & Saving
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="Churn", palette="Set2")
plt.title("Customer Churn Distribution")
plt.savefig("../visuals/churn_distribution.png", bbox_inches="tight")
plt.show()
```

```python
# Crosstab Counts & Percentage
print("--- Contract Count ---")
print(pd.crosstab(df["Contract"], df["Churn"]))

print("\n--- Contract Percentage (%) ---")
print((pd.crosstab(df["Contract"], df["Churn"], normalize="index") * 100).round(2))

# Chart
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Contract", hue="Churn", palette="coolwarm")
plt.title("Churn by Contract Type")
plt.xticks(rotation=20)
plt.savefig("../visuals/churn_by_contract.png", bbox_inches="tight")
plt.show()
```

```python
# Crosstab Counts
print("--- Internet Service Count ---")
print(pd.crosstab(df["InternetService"], df["Churn"]))

# Chart
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="InternetService", hue="Churn", palette="viridis")
plt.title("Churn by Internet Service")
plt.savefig("../visuals/churn_by_internet.png", bbox_inches="tight")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
df = pd.read_csv("../data/WA_Fn-UseC_-Telco-Customer-Churn.csv")

# Verify columns fix hua ya nahi
print("Columns loaded:", "tenure" in df.columns)

# Day 3 Analysis
df["tenure_group"] = pd.cut(df["tenure"], bins=[0, 12, 24, 48, 72], labels=["0-12", "13-24", "25-48", "49-72"])
print(pd.crosstab(df["tenure_group"], df["Churn"], normalize="index") * 100)

sns.countplot(data=df, x="tenure_group", hue="Churn")
plt.title("Churn by Tenure Group")
plt.show()
```

```python
plt.savefig("../visuals/tenure_churn_distribution.png", bbox_inches="tight")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="tenure_group", hue="Churn", palette="Set2")
plt.title("Churn by Tenure Group")
plt.xlabel("Tenure Group (Months)")
plt.ylabel("Customer Count")
```

```python
plt.savefig("../visuals/tenure_churn_distribution.png", bbox_inches="tight")
```

## Key Insights: Tenure vs Churn

- **Highest Churn in Year 1:** Substantial attrition occurs within the first 12 months (0-12 group), showing nearly equal churned vs retained customer counts.
- **Improved Long-term Retention:** As tenure increases beyond 24 months, the churn volume drops significantly.
- **Strategic Recommendation:** Focus retention campaigns and onboarding incentives on new customers during their first year of service.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Dataset load karein
df = pd.read_csv("../data/telco_customer_churn_clean.csv")
```

```python
# Monthly Charges ka statistical summary by Churn
print("--- Monthly Charges Description ---")
print(df.groupby("Churn")["MonthlyCharges"].describe().round(2))

# Monthly Charges Boxplot
plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x="Churn", y="MonthlyCharges", palette="Set2")
plt.title("Monthly Charges by Churn Status")
plt.savefig("../visuals/monthly_charges_by_churn.png", bbox_inches="tight")
plt.show()
```

```python
# Monthly Charges ko 4 groups mein divide karein
df["charge_group"] = pd.qcut(df["MonthlyCharges"], q=4, duplicates="drop")

# Charge group wise churn percentage
print("--- Churn Percentage by Charge Group (%) ---")
print((pd.crosstab(df["charge_group"], df["Churn"], normalize="index") * 100).round(2))
```

```python
# Total Charges summary by Churn
print("--- Total Charges Description ---")
print(df.groupby("Churn")["TotalCharges"].describe().round(2))
```
