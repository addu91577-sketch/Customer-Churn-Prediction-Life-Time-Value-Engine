```python
import numpy as np
import pandas as pd

# Original dataset load karein
df = pd.read_csv("../data/WA_Fn-UseC_-Telco-Customer-Churn.csv")

# Quick verification
df.head()
```

```python
# Column data types check karein
print(df.dtypes)

# TotalCharges mein blank spaces dhoondhein
blank_spaces = df[df["TotalCharges"].str.strip() == ""]
print(f"Blank rows count in TotalCharges: {len(blank_spaces)}")
```

```python
# 1. Blank spaces ko NaN mein convert karein aur numeric banayein
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"].str.strip(), errors="coerce")

# 2. Tenure 0 wale nayi customers ke TotalCharges ko 0 fill karein
df["TotalCharges"] = df["TotalCharges"].fillna(0)

# 3. Customer ID duplicates verify karein
print(f"Duplicate Customer IDs: {df['customerID'].duplicated().sum()}")

# 4. Cleaned info verify karein
df.info()
```

```python
# Cleaned CSV save karein
df.to_csv("../data/telco_customer_churn_clean.csv", index=False)
print("Cleaned CSV successfully saved at data/telco_customer_churn_clean.csv")
```
